**Travelling Salesman problem**

```python
import itertools

def solve_tsp(distance_matrix):
    num_cities = len(distance_matrix)
    cities = list(range(num_cities))
    min_distance = float('inf')
    best_path = []

    for path in itertools.permutations(cities[1:]):
        current_path = [0] + list(path) + [0]
        current_distance = 0
        for i in range(num_cities):
            current_distance += distance_matrix[current_path[i]][current_path[i+1]]
        if current_distance < min_distance:
            min_distance = current_distance
            best_path = current_path

    return best_path, min_distance

distances = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

path, distance = solve_tsp(distances)
print(f"Path: {path}")
print(f"Distance: {distance}")
```

### Graph Representation (Distance Matrix)

The problem is defined on a fully connected graph with 4 vertices (cities 0, 1, 2, 3), represented by the following distance matrix:

* **City 0 to:** City 1 (10), City 2 (15), City 3 (20)
* **City 1 to:** City 0 (10), City 2 (35), City 3 (25)
* **City 2 to:** City 0 (15), City 1 (35), City 3 (30)
* **City 3 to:** City 0 (20), City 1 (25), City 2 (30)

### Minimum Path

* **Path:** 0 → 1 → 3 → 2 → 0
* **Total Minimum Distance:** 80
