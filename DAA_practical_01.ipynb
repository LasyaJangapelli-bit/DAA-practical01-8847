```python
import time

def bubble_sort(arr):
    """
    Sorts a list using the Bubble Sort algorithm and measures execution time.
    Time Complexity: O(n^2) in the worst and average cases, O(n) in the best case.
    Space Complexity: O(1)
    """
    n = len(arr)

    start_time = time.time()

    for i in range(n):

        swapped = False
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break


    end_time = time.time()
    execution_time = end_time - start_time
    return arr, execution_time

print("Bubble Sort function defined.")
```

```python

user_input = input("Enter numbers separated by spaces to sort: ")

try:

    original_list = list(map(int, user_input.split()))


    list_to_sort = original_list[:]

    print(f"\nOriginal list: {original_list}")


    sorted_list, exec_time = bubble_sort(list_to_sort)

    print(f"Sorted list: {sorted_list}")
    print(f"Execution time: {exec_time:.6f} seconds")
    print("Time Complexity: O(n^2) for Bubble Sort (worst and average case).")
    print("This means the time taken grows quadratically with the number of elements.")

except ValueError:
    print("Invalid input. Please enter numbers separated by spaces.")
```

```python
import time

def merge_sort(arr):
    """
    Sorts a list using the Merge Sort algorithm and measures execution time.
    Time Complexity: O(n log n) in all cases (worst, average, and best).
    Space Complexity: O(n)
    """
    start_time = time.time()
    sorted_arr = _merge_sort_recursive(arr)
    end_time = time.time()
    execution_time = end_time - start_time
    return sorted_arr, execution_time

def _merge_sort_recursive(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = _merge_sort_recursive(left_half)
    right_half = _merge_sort_recursive(right_half)

    return _merge(left_half, right_half)

def _merge(left, right):
    merged_list = []
    left_idx, right_idx = 0, 0

    while left_idx < len(left) and right_idx < len(right):
        if left[left_idx] < right[right_idx]:
            merged_list.append(left[left_idx])
            left_idx += 1
        else:
            merged_list.append(right[right_idx])
            right_idx += 1


    merged_list.extend(left[left_idx:])
    merged_list.extend(right[right_idx:])

    return merged_list

print("Merge Sort function defined.")
```

```python

user_input_merge = input("Enter numbers separated by spaces to sort using Merge Sort: ")

try:

    original_list_merge = list(map(int, user_input_merge.split()))


    list_to_sort_merge = original_list_merge[:]

    print(f"\nOriginal list: {original_list_merge}")


    sorted_list_merge, exec_time_merge = merge_sort(list_to_sort_merge)

    print(f"Sorted list: {sorted_list_merge}")
    print(f"Execution time: {exec_time_merge:.6f} seconds")
    print("Time Complexity: O(n log n) for Merge Sort (all cases).")
    print("This means the time taken grows proportionally to n times the logarithm of n.")

except ValueError:
    print("Invalid input. Please enter numbers separated by spaces.")
```

## Selection Sort

```python
import time

def selection_sort(arr):
    """
    Sorts a list using the Selection Sort algorithm and measures execution time.
    Time Complexity: O(n^2) in all cases (worst, average, and best).
    Space Complexity: O(1)
    """
    n = len(arr)

    start_time = time.time()

    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i] # Swap the found minimum element with the first element

    end_time = time.time()
    execution_time = end_time - start_time
    return arr, execution_time

print("Selection Sort function defined.")
```

## Insertion Sort

```python
import time

def insertion_sort(arr):
    """
    Sorts a list using the Insertion Sort algorithm and measures execution time.
    Time Complexity: O(n^2) in the worst and average cases, O(n) in the best case.
    Space Complexity: O(1)
    """
    n = len(arr)

    start_time = time.time()

    for i in range(1, n):
        key = arr[i]
        j = i - 1
        # Move elements of arr[0..i-1], that are greater than key, to one position ahead of their current position
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

    end_time = time.time()
    execution_time = end_time - start_time
    return arr, execution_time

print("Insertion Sort function defined.")
```

## Quick Sort

```python
import time

def quick_sort(arr):
    """
    Sorts a list using the Quick Sort algorithm and measures execution time.
    Time Complexity: O(n log n) on average, O(n^2) in the worst case.
    Space Complexity: O(log n) on average, O(n) in the worst case (due to recursion stack).
    """
    start_time = time.time()
    _quick_sort_recursive(arr, 0, len(arr) - 1)
    end_time = time.time()
    execution_time = end_time - start_time
    return arr, execution_time

def _quick_sort_recursive(arr, low, high):
    if low < high:
        pi = _partition(arr, low, high)
        _quick_sort_recursive(arr, low, pi - 1)
        _quick_sort_recursive(arr, pi + 1, high)

def _partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

print("Quick Sort function defined.")
```

```python
user_input_quick = input("Enter numbers separated by spaces to sort using Quick Sort: ")

try:
    original_list_quick = list(map(int, user_input_quick.split()))
    list_to_sort_quick = original_list_quick[:]

    print(f"\nOriginal list: {original_list_quick}")

    sorted_list_quick, exec_time_quick = quick_sort(list_to_sort_quick)

    print(f"Sorted list: {sorted_list_quick}")
    print(f"Execution time: {exec_time_quick:.6f} seconds")
    print("Time Complexity: O(n log n) for Quick Sort (average case), O(n^2) (worst case).")
    print("This means the time taken grows proportionally to n times the logarithm of n on average, but can degrade to quadratic in the worst case.")

except ValueError:
    print("Invalid input. Please enter numbers separated by spaces.")
```
