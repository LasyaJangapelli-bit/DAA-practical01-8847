```python
print('**Prim\'s Algorithm**')
import sys

def prim_mst(graph):
    num_vertices = len(graph)
    selected = [False] * num_vertices
    selected[0] = True
    num_edges = 0
    mst_weight = 0

    while num_edges < num_vertices - 1:
        minimum = sys.maxsize
        x = 0
        y = 0
        for i in range(num_vertices):
            if selected[i]:
                for j in range(num_vertices):
                    if ((not selected[j]) and graph[i][j]):
                        if minimum > graph[i][j]:
                            minimum = graph[i][j]
                            x = i
                            y = j
        print(f"{x} - {y} : {graph[x][y]}")
        selected[y] = True
        mst_weight += graph[x][y]
        num_edges += 1
    print(f"Total MST Weight: {mst_weight}")

graph_prim = [
    [0, 9, 75, 0, 0],
    [9, 0, 95, 19, 42],
    [75, 95, 0, 51, 66],
    [0, 19, 51, 0, 31],
    [0, 42, 66, 31, 0]
]

prim_mst(graph_prim)
```
