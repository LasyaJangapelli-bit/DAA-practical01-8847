## Knapsack Problem

```python
def knapsack_dp(weights, values, capacity):
    """
    Solves the 0/1 Knapsack problem using dynamic programming.

    Args:
        weights (list): A list of weights for each item.
        values (list): A list of values for each item.
        capacity (int): The maximum weight the knapsack can hold.

    Returns:
        tuple: A tuple containing:
            - int: The maximum value that can be obtained.
            - list: A list of indices of items included in the knapsack.
    """
    n = len(weights)
    # dp[i][w] will store the maximum value that can be obtained
    # with first i items and a knapsack capacity of w.
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

    # Build dp table
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            # If current item's weight is more than current knapsack capacity 'w',
            # then this item cannot be included.
            if weights[i-1] > w:
                dp[i][w] = dp[i-1][w]
            else:
                # Else, consider two possibilities:
                # 1. Include the current item: values[i-1] + dp[i-1][w - weights[i-1]]
                # 2. Exclude the current item: dp[i-1][w]
                dp[i][w] = max(values[i-1] + dp[i-1][w - weights[i-1]], dp[i-1][w])

    max_value = dp[n][capacity]

    # Reconstruct the items included in the knapsack
    included_items = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            included_items.append(i-1) # Item index (0-based)
            w -= weights[i-1]
    included_items.reverse()

    return max_value, included_items

# --- User Input ---
print("Enter details for the Knapsack problem:")

num_items = int(input("Enter the number of items: "))

weights = []
values = []
for i in range(num_items):
    weight = int(input(f"Enter weight for item {i+1}: "))
    value = int(input(f"Enter value for item {i+1}: "))
    weights.append(weight)
    values.append(value)

capacity = int(input("Enter the knapsack capacity: "))

# Solve the problem
max_value, included_items_indices = knapsack_dp(weights, values, capacity)

print(f"\nMaximum value in knapsack: {max_value}")

print("Items included (0-indexed): ", end="")
if included_items_indices:
    print(', '.join(map(str, included_items_indices)))
else:
    print("None")

print("Corresponding weights and values:")
for idx in included_items_indices:
    print(f"  Item {idx} (Weight: {weights[idx]}, Value: {values[idx]})")
```
