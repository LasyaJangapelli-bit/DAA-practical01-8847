## Coin Change Algorithm

```python
def coin_change(coins, amount):
    """
    Calculates the minimum number of coins needed to make a given amount.
    Uses dynamic programming.

    Args:
        coins (list): A list of available coin denominations.
        amount (int): The target amount.

    Returns:
        int: The minimum number of coins needed, or -1 if the amount cannot be made.
    """
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for a in range(1, amount + 1):
        for coin in coins:
            if a - coin >= 0:
                dp[a] = min(dp[a], 1 + dp[a - coin])

    return dp[amount] if dp[amount] != float('inf') else -1
```

### User Input for Coin Change

```python
try:
    # Get target amount from user
    target_amount_str = input("Enter the target amount: ")
    target_amount = int(target_amount_str)

    # Get coin denominations from user
    coins_input_str = input("Enter coin denominations separated by commas or spaces (e.g., 1,2,5 or 1 2 5): ")
    # Replace commas with spaces to handle both, then split by any whitespace
    coins_processed_str = coins_input_str.replace(',', ' ')
    coins = sorted([int(c.strip()) for c in coins_processed_str.split() if c.strip()])

    if not coins:
        print("No coins entered. Please provide at least one coin denomination.")
    elif target_amount < 0:
        print("Target amount cannot be negative.")
    else:
        result = coin_change(coins, target_amount)

        if result == -1:
            print(f"It is not possible to make an amount of {target_amount} with the given coins.")
        else:
            print(f"Minimum number of coins to make {target_amount} using coins {coins}: {result}")

except ValueError:
    print("Invalid input. Please enter integers for the amount and coin denominations.")
except Exception as e:
    print(f"An error occurred: {e}")
```
